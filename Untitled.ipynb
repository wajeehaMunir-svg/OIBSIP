```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# 2. Fix Gender (convert all variations to Male or Female)
df["Gender"] = df["Gender"].replace(
    {
        "M": "Male",
        "male": "Male",
        "MALE": "Male",
        "F": "Female",
        "female": "Female",
        "FEMALE": "Female",
    }
)
```

```python
print(df["Gender"].value_counts(dropna=False), "\n")
```

```python
# 3. Clean Text Columns
for col in ["Name", "City", "Country"]:
    df[col] = df[col].astype(str).str.strip().str.title()
    df[col] = df[col].replace({"Nan": np.nan, "None": np.nan, "": np.nan})
print("Unique Cities:", df["City"].dropna().unique()[:5], "\n")
```

```python
df["email"] = (
    df["email"]
    .astype(str)
    .str.strip()
    .str.lower()
    .apply(lambda x: x if "@" in str(x) else np.nan)
)
df["Phone_number"] = (
    df["Phone_number"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)
    .replace({"0": np.nan, "": np.nan})
)
df["CustomerID"] = (
    df["CustomerID"].astype(str).str.extract(r"(\d+)")[0].fillna("Unknown")
)
```

```python
4.# Remove bad placeholder numbers
df["Age"] = df["Age"].replace([-5, 200], np.nan)
df["purchase_amount"] = df["purchase_amount"].replace(-999, np.nan)
df["feedback_score"] = df["feedback_score"].replace(-1, np.nan)
print("Missing numeric values after cleaning placeholders:")
print(df[["Age", "purchase_amount", "feedback_score"]].isnull().sum(), "\n")
```

```python
# 5. Convert date columns
df["Signup_Date"] = pd.to_datetime(
    df["Signup_Date"], errors="coerce", dayfirst=True
)
df["Last_purchase_date"] = pd.to_datetime(
    df["Last_purchase_date"], errors="coerce", dayfirst=True
)

print(df[["Signup_Date", "Last_purchase_date"]].dtypes, "\n")
```

```python
# 6. Fill missing numbers with Median
df["Age"] = df["Age"].fillna(df["Age"].median())
df["purchase_amount"] = df["purchase_amount"].fillna(
    df["purchase_amount"].median()
)
df["feedback_score"] = df["feedback_score"].fillna(df["feedback_score"].median())
```

```python
# Outlier Capping via IQR
for col in ["Age", "purchase_amount", "feedback_score"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    df[col] = np.clip(df[col], Q1 - 1.5 * IQR, Q3 + 1.5 * IQR)
```

```python
# 7. Fill missing text with Mode
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])
df["Name"] = df["Name"].fillna(df["Name"].mode()[0])

print("Remaining Nulls in Key Columns:", df[["Gender", "Age", "City"]].isnull().sum().sum(), "\n")
```

```python
summary_table = pd.DataFrame(
    {
        "Column": df_raw.columns,
        "Before Nulls": df_raw.isnull().sum().values,
        "After Nulls": df.isnull().sum().values,
        "Before Dtype": df_raw.dtypes.values,
        "After Dtype": df.dtypes.values,
    }
)
display(summary_table)
```
