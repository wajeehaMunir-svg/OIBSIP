```python
# 3. Handle missing values and inconsistent data
# Drop rows without CustomerID
df_clean = df.dropna(subset=['CustomerID']).copy()
```

```python
df_clean = df_clean[(df_clean['Quantity'] > 0) & (df_clean['UnitPrice'] > 0)]
```

```python
# Format data types
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
df_clean['TotalSpend'] = df_clean['Quantity'] * df_clean['UnitPrice']
```

```python

print("Cleaned Dataset Shape:", df_clean.shape)
```

```python
##Create RFM features
```

<div style="background: linear-gradient(135deg, #a8edd8 0%, #a8d8ea 50%, #aa96da 100%); padding: 22px 28px; border-radius: 18px; color: #2d3748; margin-top: 20px; margin-bottom: 25px; box-shadow: 0px 4px 15px rgba(170, 150, 218, 0.3); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; display: flex; justify-content: space-between; align-items: center; flex-wrap: wrap; gap: 15px;">
  <div>
    <h2 style="margin: 0; font-size: 26px; font-weight: 800; color: #1a202c; letter-spacing: -0.5px;">
      👥 RFM Customer Analysis
    </h2>
    <p style="margin: 8px 0 0 0; font-size: 15px; font-weight: 700; color: #2d3748;">
      🕒 Recency &nbsp;|&nbsp; 🔄 Frequency &nbsp;|&nbsp; 💰 Monetary &nbsp;|&nbsp; 📈 RFM Descriptive Statistics
    </p>
  </div>
</div>

```python
reference_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1)
```

```python
rfm = df_clean.groupby('CustomerID').agg(
    Recency=('InvoiceDate', lambda x: (reference_date - x.max()).days),
    Frequency=('InvoiceNo', 'nunique'),
    Monetary=('TotalSpend', 'sum')
).reset_index()
```

```python
rfm.head()
```

```python
average_purchase_value = (
    df_clean['TotalSpend'].sum() /
    df_clean['InvoiceNo'].nunique()
)
```

```python
print("Average Purchase Value:", round(average_purchase_value, 2))
```

```python
purchase_frequency = (
    df_clean.groupby('CustomerID')['InvoiceNo']
    .nunique()
    .mean()
)
```

```python
print("Average Purchase Frequency:", round(purchase_frequency, 2))
```

```python
customer_lifetime_value = (
    df_clean.groupby('CustomerID')['TotalSpend']
    .sum()
    .mean()
)
```

```python
print("Average Customer Lifetime Value:", round(customer_lifetime_value, 2))
```

```python
### RFM descriptive statistics
print(rfm[['Recency', 'Frequency', 'Monetary']].describe())
```

      ⚙️ K-Means Preparation
    </h2>
    <p style="margin: 8px 0 0 0; font-size: 15px; font-weight: 700; color: #2d3748;">
      🎯 Select RFM Features &nbsp;|&nbsp; ⚖️ StandardScaler &nbsp;|&nbsp; 📊 Scaled RFM Data
    </p>

```python
##Select RFM features 
features = rfm[['Recency', 'Frequency', 'Monetary']]
```

```python
##Standardise the data
scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(features)

rfm_scaled = pd.DataFrame(
    rfm_scaled,
    columns=['Recency', 'Frequency', 'Monetary']
)

rfm_scaled.head()
```

<div style="background: linear-gradient(135deg, #a8edd8 0%, #a8d8ea 50%, #aa96da 100%); padding: 22px 28px; border-radius: 18px; color: #2d3748; margin-top: 20px; margin-bottom: 25px; box-shadow: 0px 4px 15px rgba(170, 150, 218, 0.3); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; display: flex; justify-content: space-between; align-items: center; flex-wrap: wrap; gap: 15px;">
  <div>
    <h2 style="margin: 0; font-size: 26px; font-weight: 800; color: #1a202c; letter-spacing: -0.5px;">
      📈 Determine Optimal Clusters
    </h2>
    <p style="margin: 8px 0 0 0; font-size: 15px; font-weight: 700; color: #2d3748;">
      📉 Elbow Method &nbsp;|&nbsp; 🎯 Best K Selection
    </p>
  </div>
</div>

```python
inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```

```python
K = 4
```

```python
##Apply K-Means
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm['Cluster'] = kmeans.fit_predict(rfm_scaled)

rfm.head()
```

<div style="background: linear-gradient(135deg, #a8edd8 0%, #a8d8ea 50%, #aa96da 100%); padding: 22px 28px; border-radius: 18px; color: #2d3748; margin-top: 20px; margin-bottom: 25px; box-shadow: 0px 4px 15px rgba(170, 150, 218, 0.3); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; display: flex; justify-content: space-between; align-items: center; flex-wrap: wrap; gap: 15px;">

    <p style="margin: 8px 0 0 0; font-size: 15px; font-weight: 700; color: #2d3748;">
      🤖 K-Means Clusters &nbsp;|&nbsp; 🏷️ Meaningful Cluster Names &nbsp;|&nbsp; 👤 Customer Type Profile
    </p>
  </div>

```python
##Profile Each Cluster
cluster_profile = rfm.groupby('Cluster').agg(
    Customers=('CustomerID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
).round(2)

cluster_profile
```

```python
###Give Clusters Meaningful Names
cluster_profile = cluster_profile.sort_values(
    'Avg_Monetary',
    ascending=False
)

cluster_names = {
    cluster_profile.index[0]: 'High-Value Customers',
    cluster_profile.index[1]: 'Regular Customers',
    cluster_profile.index[2]: 'Occasional Customers',
    cluster_profile.index[3]: 'Low-Value / At-Risk Customers'
}

rfm['CustomerType'] = rfm['Cluster'].map(cluster_names)

cluster_profile
```

```python
### Customer Type Profile
profile = rfm.groupby('CustomerType').agg(
    Customers=('CustomerID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
).round(2)

profile
```
